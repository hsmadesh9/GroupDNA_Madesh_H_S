GroupDNA is a WhatsApp group chat analysis project developed using Python fundamentals and NumPy.

```python
def parse_chat(content):
    messages = []
    system_count = 0
    media_count = 0
    deleted_count = 0

    lines = content.split("\n")

    for line in lines:
        if line.strip() == "":
            continue

        if " - " in line:
            parts = line.split(" - ", 1)
            timestamp = parts[0]
            remaining = parts[1]

            if ": " in remaining:
                sender_message = remaining.split(": ", 1)
                sender = sender_message[0]
                message = sender_message[1]

                if message == "This message was deleted":
                    deleted_count = deleted_count + 1

                if "<Media omitted>" in message:
                    media_count = media_count + 1

                messages.append({
                    "timestamp": timestamp,
                    "sender": sender,
                    "message": message
                })

            else:
                system_count = system_count + 1

        else:
            if len(messages) > 0:
                messages[-1]["message"] = messages[-1]["message"] + " " + line

    return messages, system_count, media_count, deleted_count


messages, system_count, media_count, deleted_count = parse_chat(sample_chat)

print("Total messages:", len(messages))
print("System messages:", system_count)
print("Media messages:", media_count)
print("Deleted messages:", deleted_count)

print("\nFirst few messages:")
for item in messages[:5]:
    print(item)
```

```python
for item in messages:
    item["datetime"] = datetime.strptime(item["timestamp"], "%d/%m/%y, %H:%M")

participants = set()

for item in messages:
    participants.add(item["sender"])

print("Total participants:", len(participants))

message_count = {}

for person in participants:
    message_count[person] = 0

for item in messages:
    name = item["sender"]
    message_count[name] = message_count[name] + 1

dates = []

for item in messages:
    dates.append(item["datetime"].date())

first_date = min(dates)
last_date = max(dates)

print("\nGROUP OVERVIEW")
print("-------------------------")
print("Total messages:", len(messages))
print("Total participants:", len(participants))
print("First message date:", first_date)
print("Last message date:", last_date)

print("\nMessages sent by each person:")

for person in participants:
    print(person, ":", message_count[person])
```

```python
day_count = {}
hour_count = {}

for item in messages:
    day = item["datetime"].date()
    hour = item["datetime"].hour

    if day in day_count:
        day_count[day] = day_count[day] + 1
    else:
        day_count[day] = 1

    if hour in hour_count:
        hour_count[hour] = hour_count[hour] + 1
    else:
        hour_count[hour] = 1

busiest_day = max(day_count, key=day_count.get)
active_hour = max(hour_count, key=hour_count.get)

print("MOST ACTIVE DAY AND HOUR")
print("-------------------------")

print("Busiest day:", busiest_day)
print("Number of messages:", day_count[busiest_day])

print("Most active hour:", active_hour, ":00")
print("Number of messages:", hour_count[active_hour])
```

```python
word_count = {}

for item in messages:
    message = item["message"]

    words = message.lower().split()

    for word in words:
        word = word.strip(".,!?():;\"'")

        if word != "":
            if word in word_count:
                word_count[word] = word_count[word] + 1
            else:
                word_count[word] = 1

top_words = sorted(word_count, key=word_count.get, reverse=True)

print("TOP 10 MOST USED WORDS")
print("-------------------------")

count = 0

for word in top_words:
    print(word, ":", word_count[word])
    count = count + 1

    if count == 10:
        break
```

```python
response_times = {}
previous_message = {}

for item in messages:
    person = item["sender"]
    current_time = item["datetime"]

    if person in previous_message:
        time_gap = current_time - previous_message[person]

        minutes = time_gap.total_seconds() / 60

        if person in response_times:
            response_times[person].append(minutes)
        else:
            response_times[person] = [minutes]

    previous_message[person] = current_time

print("RESPONSE TIME")
print("-------------------------")

for person in response_times:
    total = sum(response_times[person])
    count = len(response_times[person])
    average = total / count

    print(person, "average time between messages:", round(average, 2), "minutes")


messages.sort(key=lambda item: item["datetime"])

longest_gap = timedelta(0)

for i in range(1, len(messages)):
    time_gap = messages[i]["datetime"] - messages[i - 1]["datetime"]

    if time_gap > longest_gap:
        longest_gap = time_gap

print("\nSILENT STREAK")
print("-------------------------")
print("Longest time without a message:", longest_gap)
```

```python
person_activity = {}

for person in participants:
    person_activity[person] = {
        "messages": 0,
        "night_messages": 0,
        "total_words": 0
    }

for item in messages:
    person = item["sender"]
    hour = item["datetime"].hour
    message = item["message"]

    person_activity[person]["messages"] = person_activity[person]["messages"] + 1

    if hour >= 22 or hour < 5:
        person_activity[person]["night_messages"] = person_activity[person]["night_messages"] + 1

    words = message.split()
    person_activity[person]["total_words"] = person_activity[person]["total_words"] + len(words)

print("PERSONALITY ARCHETYPES")
print("-------------------------")

for person in person_activity:
    total_messages = person_activity[person]["messages"]
    night_messages = person_activity[person]["night_messages"]
    total_words = person_activity[person]["total_words"]

    average_words = total_words / total_messages

    if total_messages == max(message_count.values()):
        personality = "Talkative Person"

    elif night_messages > total_messages / 2:
        personality = "Night Owl"

    elif average_words > 10:
        personality = "Detailed Chatter"

    else:
        personality = "Silent Observer"

    print(person, ":", personality)
```
